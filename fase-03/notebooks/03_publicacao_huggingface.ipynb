Setup


```python
from google.colab import drive
drive.mount("/content/drive")

REPO_ID = "FernandoMarquesVale/Llama-3.2-3B-medassist-ptbr"
GGUF_PATH = "/content/drive/MyDrive/fiap_fase3/Llama-3.2-3B-Instruct.Q4_K_M.gguf"
MODELFILE_PATH = "/content/drive/MyDrive/fiap_fase3/Modelfile"
CARD_PATH = "/content/drive/MyDrive/fiap_fase3/README.md"
LORA_DIR = "/content/drive/MyDrive/fiap_fase3/lora_v2"

import os
for p in [GGUF_PATH, MODELFILE_PATH, CARD_PATH]:
    print(("OK   " if os.path.exists(p) else "FALTA"), p, f"{os.path.getsize(p)/1e9:.2f} GB" if os.path.exists(p) else "")
```

Autenticação

```python
from google.colab import userdata
from huggingface_hub import login, whoami
login(token=userdata.get("HF_TOKEN"))
print("Logado como:", whoami()["name"])
```

Repositório


```python
from huggingface_hub import HfApi
api = HfApi()
api.create_repo(REPO_ID, repo_type="model", private=False, exist_ok=True)

api.upload_file(path_or_fileobj=CARD_PATH, path_in_repo="README.md", repo_id=REPO_ID,
                commit_message="Model card")
api.upload_file(path_or_fileobj=MODELFILE_PATH, path_in_repo="Modelfile", repo_id=REPO_ID,
                commit_message="Modelfile para Ollama")
api.upload_file(path_or_fileobj=GGUF_PATH, path_in_repo="Llama-3.2-3B-medassist-ptbr.Q4_K_M.gguf",
                repo_id=REPO_ID, commit_message="GGUF q4_k_m (fine-tuning v2, formato RAG)")
if LORA_DIR:
    api.upload_folder(folder_path=LORA_DIR, path_in_repo="lora_adapter", repo_id=REPO_ID,
                      commit_message="Adaptador LoRA v2")
print("Publicado em: https://huggingface.co/" + REPO_ID)
```
